```python
gajiPerjam = int(input("masukkan gaji perjam Budi:"))
jamKerja = int(input('masukkan jam kerja Budi:'))

gajiBudi = ((jamKerja * gajiPerjam) * 5)
GajiPajak = (gajiBudi - ((14/100)*gajiBudi))
BeliBaju = ((10/100)*GajiPajak)
AlatTulis = ((1/100)*GajiPajak)
sedekah = ((25/100)*GajiPajak)
yatim = (sedekah * (30/100))
dhuafa = sedekah - yatim
print('Pendapatan Budi selama libur musim panas sebelum melakukan pembayaran pajak:', gajiBudi)
print('Pendapatan Budi selama libur musim panas setelah melakukan pembayaran pajak:', GajiPajak)
print('Jumlah uang yang akan Budi habiskan untuk membeli pakaian dan aksesoris:', BeliBaju)
print('Jumlah uang yang akan Budi habiskan untuk membeli alat tulis:', AlatTulis)
print('Jumlah uang yang akan Budi sedekahkan:', sedekah)
print('Jumlah uang yang akan diterima anak yatim:', yatim)
print('Jumlah uang yang akan diterima kaum dhuafa:', dhuafa)
```
